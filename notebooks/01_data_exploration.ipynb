## Basic Student Behaviour Aggregation

```python
basic_activity = (
    student_vle_sample
    .groupby(
        [
            "id_student",
            "code_module",
            "code_presentation"
        ]
    )
    .agg(
        total_clicks=("sum_click", "sum"),
        active_days=("date", "nunique"),
        first_activity=("date", "min"),
        last_activity=("date", "max"),
        unique_resources=("id_site", "nunique")
    )
    .reset_index()
)

basic_activity.head()
```

```python
basic_activity.shape
```

```python
basic_activity["activity_span"] = (
    basic_activity["last_activity"]
    - basic_activity["first_activity"]
)

basic_activity.head()
```

```python
student_vle_sample["week"] = (
    student_vle_sample["date"] // 7
)
```

```python
active_weeks = (
    student_vle_sample
    .groupby(
        [
            "id_student",
            "code_module",
            "code_presentation"
        ]
    )["week"]
    .nunique()
    .reset_index(name="active_weeks")
)

active_weeks.head()
```

```python
basic_activity = basic_activity.merge(
    active_weeks,
    on=[
        "id_student",
        "code_module",
        "code_presentation"
    ],
    how="left"
)

basic_activity.head()
```

```python
student_vle = pd.read_csv(
    RAW_DIR / "studentVle.csv",
    dtype={
        "code_module": "category",
        "code_presentation": "category",
        "id_student": "int32",
        "id_site": "int32",
        "date": "int16",
        "sum_click": "int32"
    }
)

print("Loaded successfully")
```

```python
student_vle.shape
```

```python
student_vle.info(memory_usage="deep")
```

```python
print("Rows:", len(student_vle))
print("Unique students:", student_vle["id_student"].nunique())
print("Unique resources:", student_vle["id_site"].nunique())
print("Minimum date:", student_vle["date"].min())
print("Maximum date:", student_vle["date"].max())
print("Total clicks:", student_vle["sum_click"].sum())
```

```python
student_vle.head()
```

```python
student_vle["week"] = student_vle["date"] // 7
```

```python
student_activity = (
    student_vle
    .groupby(
        [
            "id_student",
            "code_module",
            "code_presentation"
        ],
        observed=True
    )
    .agg(
        total_clicks=("sum_click", "sum"),
        active_days=("date", "nunique"),
        active_weeks=("week", "nunique"),
        first_activity=("date", "min"),
        last_activity=("date", "max"),
        unique_resources=("id_site", "nunique")
    )
    .reset_index()
)

student_activity.head()
```

```python
student_activity["activity_span"] = (
    student_activity["last_activity"]
    - student_activity["first_activity"]
)
```

```python
student_activity.shape
```

```python
student_activity.describe()
```

```python
activity_with_type = student_vle.merge(
    vle[
        [
            "code_module",
            "code_presentation",
            "id_site",
            "activity_type"
        ]
    ],
    on=[
        "code_module",
        "code_presentation",
        "id_site"
    ],
    how="left"
)
```

```python
activity_with_type[
    [
        "id_student",
        "date",
        "sum_click",
        "activity_type"
    ]
].head()
```

```python
activity_with_type["activity_type"].value_counts()
```

```python
activity_type_summary = (
    activity_with_type
    .groupby(
        [
            "id_student",
            "code_module",
            "code_presentation",
            "activity_type"
        ],
        observed=True
    )["sum_click"]
    .sum()
    .reset_index(name="activity_clicks")
)

activity_type_summary.head()
```

```python
preferred_activity = (
    activity_type_summary
    .sort_values(
        "activity_clicks",
        ascending=False
    )
    .drop_duplicates(
        subset=[
            "id_student",
            "code_module",
            "code_presentation"
        ]
    )
)
```

```python
preferred_activity = preferred_activity[
    [
        "id_student",
        "code_module",
        "code_presentation",
        "activity_type"
    ]
].rename(
    columns={
        "activity_type": "preferred_activity"
    }
)

preferred_activity.head()
```

```python
student_activity = student_activity.merge(
    preferred_activity,
    on=[
        "id_student",
        "code_module",
        "code_presentation"
    ],
    how="left"
)

student_activity.head()
```

```python
student_activity.to_csv(
    PROCESSED_DIR / "student_activity_summary.csv",
    index=False
)
```
